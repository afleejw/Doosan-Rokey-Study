아이디어
1. 최대 값 찾기 및 개수 확인
2. 원본 리스트 0~쵀대 값 수로 슬라이싱
- 만약 최대값 수가 교환 횟수 보다 크으면 교환 횟수 사용
3. 슬라이싱 리스트에 최대값 있으면 제거
4. 슬라이싱 리스트 내림차순 정렬
5. 최대값 위치 리스트 생성: [list.index(max, i) for i in range(max count)]
6. 정렬된 리스트와 최대값 위치 리스트를 각각 pop함
7. 정렬된 리스트에서 pop한 값의 원본 위치와 pop한 최대값 위치 교환
8. 교환 횟수 차감
9. 교환 부위 제외한 나머지 반복
- 만약 나머지 부분의 수가 2이면 횟수 % 2의 값으로 교환

```python
case_num = int(input())
test_data = [input().strip().split() for i in range(case_num)]

for num, test in enumerate(test_data):

  idx = 0
  count = 0
  data = list(map(int, list(test[0])))

  while count < int(test[1]):
    # 1. 최대 값 찾기 및 개수 확인
    max_num = max(data[idx:])
    max_count = data[idx:].count(max_num)

    # 2. 원본 리스트 슬라이싱
    slice_point = max_count
    if max_count > (int(test[1]) - count):
      slice_point = (int(test[1]) - count)

    # 3. 슬라이싱 리스트 최대값 제거
    replace_list = data[idx:slice_point]
    for i in range(replace_list.count(max_num)):
      replace_list.remove(max_num)

    # 4. 슬라이싱 리스트 내림차순 정렬
    replace_list = sorted(replace_list, reverse=True)

    # 5. 최대값 위치 리스트 생성
    max_point_list = [i for i, value in enumerate(data[idx:]) if value == max_num]
    print(max_point_list)

    # 6. 정렬된 리스트와 최대값 위치 리스트를 각각 pop함
    # 7. 정렬된 리스트에서 pop한 값의 원본 위치와 pop한 최대값 위치 교환
    # 8. 교환 횟수 차감
    for i in range(len(replace_list)):
      max_point = max_point_list.pop()
      replace_num = replace_list.pop()
      replace_point = data.index(replace_num)
      data[max_point], data[replace_point] = data[replace_point], data[max_point]
      count += 1

    # 9. 교환 부위 제외한 나머지 반복
    idx += max_count
    if len(data[idx:]) == 2:
      if (int(test[1]) - count) % 2 == 1:
        data[-2], data[-1] = data[-1], data[-2]
      break

  data = ''.join(list(map(str, data)))
  print(f'#{num+1} {data}')
```

백트래킹 방식

```python
def dfs(arr, cnt, result):
  if cnt == 0:
    num = int(''.join(arr))
    result = max(result, num)
    return result

  for i in range(len(arr)):
    for j in range(i+1, len(arr)):
      arr[i], arr[j] = arr[j], arr[i]
      temp = ''.join(arr)
      if (temp, cnt-1) not in visited:
        visited.add((temp, cnt-1))
        result = dfs(arr, cnt-1, result)
      arr[i], arr[j] = arr[j], arr[i]
  return result

case_num = int(input())
test_data = [input().strip().split() for i in range(case_num)]

for num, test in enumerate(test_data):

  numbers, count = test[0], int(test[1])
  data = list(numbers)

  visited = set()
  result = dfs(data, count, 0)

  print(f'#{num+1} {result}')
```
