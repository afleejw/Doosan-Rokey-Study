```python
import sys
input = sys.stdin.read

# 방향 저장 (북, 동, 남, 서)
dx = [-1, 0, 1, 0]  # x축 이동 (행)
dy = [0, 1, 0, -1]  # y축 이동 (열)

def clean_room(n, m, r, c, d, room):
    # 청소한 구역 수를 생성
    cleaned_count = 0

    while True:
        # 1. 현재 위치를 청소
        if room[r][c] == 0:  # 청소되지 않은 칸이였을때 if문으로 가정
            room[r][c] = 2  # 청소 완료 표시하였을 시  (2)로표기
            cleaned_count += 1

        # 2. 현 위치에서 4방향 확인
        found_cleanable = False
        for _ in range(4):
            d = (d - 1) % 4  # 반시계 방향으로 회전
            nx, ny = r + dx[d], c + dy[d]

            if 0 <= nx < n and 0 <= ny < m and room[nx][ny] == 0:  # 청소할 수 있는 칸이면 가정으로진행
                r, c = nx, ny  # 앞으로
                found_cleanable = True
                break

        if not found_cleanable:  # 네 방향 모두 청소할 곳이 없으면 가정으로 상황을 설정
            back_x, back_y = r - dx[d], c - dy[d]  # 뒤로
            if 0 <= back_x < n and 0 <= back_y < m and room[back_x][back_y] != 1:  # 후진 가능
                r, c = back_x, back_y
            else:  # 후진 불가능 (벽혹은 장애물이 있으면 종료)
                return cleaned_count

# 입력 처리
data = input().split('\n')
n, m = map(int, data[0].split())  # 방 크기
r, c, d = map(int, data[1].split())  # 로봇 처음 위치 및 방향
room = [list(map(int, row.split())) for row in data[2:n+2]]  # 방 정보

# 결과 출력
print(clean_room(n, m, r, c, d, room))
```
